```python
data.isnull().sum()
```

AS the cabin has missing values of around 687 that is about 77% of the dataset, filling it also create a artificial bias of the dataset, hence dropping it

```python
data = data.drop(columns=['Cabin'])
```

```python
data.head()
```

filling the age null values with median values , as the age is continuous and may have outliers , so using median instead of mean

```python
data['Age'] = data['Age'].fillna(data['Age'].median())
```

```python
data.isnull().sum()
```

filling the embarked null values with mode, as it is a categorical column filling with the most frequent one

```python
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data['Sex'].value_counts()
```

```python
data['Sex'] = data['Sex'].replace({'male':0,"female":1})
```

```python
data.head()
```

```python
data.info()
```

droping the name, ticket, passengerId as they are of no use in model learning

```python
data.drop(columns=['Name','PassengerId','Ticket'], axis=1, inplace=True)
```

```python
data.head()
```

```python
data.info()
```

```python
data.nunique()
```

```python
data = pd.get_dummies(data, columns=['Embarked'], drop_first=True)
```

```python
data.head()
```

```python
data.info()
```

```python
x = data.drop(columns=['Survived'], axis=1)
y = data['Survived']
```

```python
print(x)
print(y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
DT = DecisionTreeClassifier(max_depth=3)
DT.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
y_pred = DT.predict(x_test)
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

The model achieves an overall accuracy of ~82%, with strong performance on class 0 and slightly lower performance on class 1. While errors are relatively balanced, the model shows mild bias toward the majority class. The recall for class 1 (76%) indicates that some positive cases are still being missed. Overall, the model performs reasonably well but could be improved for better detection of class 1 instances.
